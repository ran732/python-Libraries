# Selecting rows from a DataFrame

* iloc - searches using index positions
* loc - searches using index labels

* **iloc vs loc  -> while slicing**

> iloc don't include last [1:4], not 4 incuded

> loc include last [1:4], 4 included

```python
# Action movies rating higher than 7.5

movies[(movies['genres'].str.split('|').apply(lambda x:'Action' in x))  & (movies['imdb_rating']>7.5)].shape[0]

movies[(movies['genres'].str.contains('Action'))  & (movies['imdb_rating']>7.5)].shape[0]
```

```python
def check_action(genre_list):

    if 'Action' in genre_list:
        return True
    return False


movies['split_genres'] = movies['genres'].str.split("|")

action_movies = movies[movies['split_genres'].apply(check_action) & (movies['imdb_rating']>7.5)].shape[0]


action_movies
```
